```python
#Create List
Student_list = ["Lewis", "Max", "Sebastian", "Ayrton"]

print("\nStudent List:")
print(Student_list)

#Create Tuple
Student_tuple = ("Lewis", "Max", "Sebastian", "Ayrton")

print("\nStudent Tuple:")
print(Student_tuple)

#Create Dictionary
Student_dict = {
 1:"Lewis",
 2:"Max" ,
 3:"Sebastian",
 4:"Ayrton",
    }
print("\nStudent Dictionary:")
print(Student_dict)

#Adding in Dictionary
Student_dict[5]="Fernando"

print("\nUpdated Student Dictionary:")
print(Student_dict)

#Replaceing names
Student_dict[2]="Daniel"

print("\nUpdated Student Dictionary:")
print(Student_dict)

#Delete a student
del Student_dict[5]

print("\nUpdated Student Dictionary:")
print(Student_dict)
```
